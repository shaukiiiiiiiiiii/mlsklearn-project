```python
df.drop(columns=['Cabin'], inplace=True)
```

```python
sns.boxenplot(x='Pclass', y='Age', data=df)
plt.title('Age Distribution by Pclass')
plt.xlabel('Pclass')
plt.ylabel('Age')
plt.legend(title='Pclass')
plt.show()
```

```python
figsize = (10, 6)
plt.figure(figsize=figsize)
pallette = sns.color_palette("Set2")
sns.scatterplot(x='Age', y='Fare', hue='Survived', data=df, palette=pallette, alpha=0.7)
plt.title('Age vs Fare by Survived')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid()
plt.legend(title='Survived')
plt.savefig('age_vs_fare_by_survived.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
columns = ["Sex", "Pclass", "Embarked", "SibSp"]

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for ax, column in zip(axes, columns):
    sns.countplot(
        data=df,
        x=column,
        hue="Survived",
        palette=pallette[:2],
        linewidth=1.2,
        edgecolor="white",
        ax=ax
    )

    # Style bar outlines and chart
    for bar in ax.patches:
        bar.set_linestyle("-")

    ax.set_title(f"Passenger Count by {column}", fontsize=13, weight="bold")
    ax.set_xlabel(column, fontsize=11)
    ax.set_ylabel("Passenger Count", fontsize=11)
    ax.grid(axis="y", linestyle="--", linewidth=0.7, alpha=0.5)
    ax.set_axisbelow(True)
    ax.legend(title="Survived", labels=["No", "Yes"], frameon=False)

fig.suptitle("Titanic Passenger Categories and Survival", fontsize=17, weight="bold")
fig.tight_layout()
fig.savefig("titanic_category_bar_charts.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
sns.barplot(
    data=df,
    x="Pclass",
    y="Survived",
    errorbar=None,
    color=pallette[1]  # changed color
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.ylim(0, 1)

plt.savefig("survival_rate_by_pclass.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
age_bins = [0, 12, 18, 30, 45, 60, np.inf]
age_labels = ["0–12", "13–18", "19–30", "31–45", "46–60", "61+"]

plot_df = df.assign(
    AgeGroup=pd.cut(df["Age"], bins=age_bins, labels=age_labels)
)
survival_by_age = (
    plot_df.groupby(["AgeGroup", "Sex"], observed=False)["Survived"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(10, 6))
sns.lineplot(
    data=survival_by_age,
    x="AgeGroup",
    y="Survived",
    hue="Sex",
    hue_order=["female", "male"],
    palette={"female": "#e76f51", "male": "#457b9d"},
    marker="o",
    linewidth=2.5,
)

plt.title("Titanic Survival Rate by Age Group and Sex")
plt.xlabel("Age group")
plt.ylabel("Survival rate")
plt.ylim(0, 1)
plt.legend(title="Sex")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.savefig("survival_rate_by_age_and_sex.png", dpi=300, bbox_inches="tight")
plt.show()
```
